```python
from udemy_purchase_scraper.utils import loadOwned, loadPurchases
from udemy_purchase_scraper.purchase.classes import UdemyPurchase
from udemy_purchase_scraper.owned.classes import UdemyCourseOwn
```

```python
owned = loadOwned()
purchases = loadPurchases()
```

```python
def toDict(purchase: UdemyPurchase):
    """
    x
    """
    return({
        "purchase_number": purchase.purchase_number,
        "purchase_date": purchase.purchase_date,
        "purchase_cost": purchase.purchase_cost
    })
```

```python
toDict(purchases[0])
```

```python
def toDict(course: UdemyCourseOwn):
    """
    x
    """
    return({
        "course_name": course.course_name,
        "course_author": course.course_author,
        "cost": course.purchase.course_cost,
        "purchase_date": course.purchase.course_date,
        "purchase_number": course.purchase.course_purchase
    })
```
